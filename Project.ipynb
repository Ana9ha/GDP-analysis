```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the data
df = pd.read_csv("GDP_2.csv")

# Create the heatmap
plt.figure(figsize=(10, 7))
sns.heatmap(df.pivot_table(values='Unemployment percentage', index='Year', columns='Country'), annot=True, cmap='twilight', fmt=".2f")
plt.title('Unemployment Percentage Heatmap by Country and Year')
plt.xlabel('Country')
plt.ylabel('Year')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
india_data = df[df['Country'] == 'China']
total_population = india_data['Population, total'].iloc[0]
unemployment_percentage = india_data['Unemployment percentage'].iloc[0]
unemployed_people = (unemployment_percentage / 100) * total_population
labels = ['Employed', 'Unemployed']
sizes = [total_population - unemployed_people, unemployed_people]
colors = ['pink', 'lightgreen']
plt.figure(figsize=(8, 8))
plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
plt.title('China Population: Employed vs Unemployed')
plt.axis('equal')
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
df_india = pd.read_csv("India_GDP_1960-2022.csv")

df_india['India GDP - Historical Data'] = pd.to_numeric(df_india['India GDP - Historical Data'], errors='coerce')

df_limited = df_india[df_india['India GDP - Historical Data'] >= 2000]

df_reversed = df_limited.iloc[::-1]

plt.figure(figsize=(12, 6))
plt.fill_between(df_reversed['India GDP - Historical Data'], df_reversed['India GDP - Historical Data.3'], color="pink", alpha=0.4)
plt.plot(df_reversed['India GDP - Historical Data'], df_reversed['India GDP - Historical Data.3'], color="green", alpha=0.6, linewidth=2)
plt.title("India's GDP percentage over time")
plt.xlabel('Year')
plt.ylabel('GDP growth percentage')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
